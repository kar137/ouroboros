## Verify GPU Availability

```python
import torch

try:
    # Check if CUDA (GPU support) is available
    if torch.cuda.is_available():
        print("✅ GPU is available and ready for use!\n")
        
        # Get the number of available GPUs
        num_gpus = torch.cuda.device_count()
        print(f" Number of GPUs available: {num_gpus}\n")
        
        # Get detailed information about each GPU
        for i in range(num_gpus):
            gpu_name = torch.cuda.get_device_name(i)
            gpu_memory = torch.cuda.get_device_properties(i).total_memory / (1024**3)  # Convert to GB
            print(f"GPU {i}: {gpu_name}")
            print(f"  Total Memory: {gpu_memory:.2f} GB")
        
        # Display CUDA version
        cuda_version = torch.version.cuda
        print(f"\n CUDA Version: {cuda_version}")
        
    else:
        print("⚠️ WARNING: GPU is NOT available!")
        print("\nPossible reasons:")
        print("  1. You haven't selected a GPU runtime")
        print("  2. Go to Runtime → Change runtime type → Select T4 GPU")
        print("  3. GPU quota might be exhausted (try again later)")
        print("\n⚠️ Training will run on CPU, which will be significantly slower.")
        
except Exception as e:
    print(f"❌ Error checking GPU availability: {e}")
```

## System Information Utility Function


```python
def print_system_info():
    import torch
    import sys
    
    print("="*60)
    print("SYSTEM INFORMATION")
    print("="*60)
    
    # Python version
    print(f"\n Python Version: {sys.version.split()[0]}")
    
    # PyTorch version
    print(f" PyTorch Version: {torch.__version__}")
    
    # CUDA information
    if torch.cuda.is_available():
        print(f"\n CUDA Available: Yes")
        print(f" CUDA Version: {torch.version.cuda}")
        print(f" cuDNN Version: {torch.backends.cudnn.version()}")
        print(f"\n GPU Device Count: {torch.cuda.device_count()}")
        
        # Information for each GPU
        for i in range(torch.cuda.device_count()):
            print(f"\n  GPU {i}: {torch.cuda.get_device_name(i)}")
            
            # Memory information
            total_memory = torch.cuda.get_device_properties(i).total_memory / (1024**3)
            allocated_memory = torch.cuda.memory_allocated(i) / (1024**3)
            cached_memory = torch.cuda.memory_reserved(i) / (1024**3)
            
            print(f"    Total Memory: {total_memory:.2f} GB")
            print(f"    Allocated Memory: {allocated_memory:.2f} GB")
            print(f"    Cached Memory: {cached_memory:.2f} GB")
            print(f"    Available Memory: {total_memory - allocated_memory:.2f} GB")
    else:
        print(f"\n CUDA Available: No")
        print("  ⚠️ Running on CPU - training will be slower")
    
    print("\n" + "="*60)

# Call the function to display system information
print_system_info()
```

## GPU Test

```python
import torch
import time

try:
    print(" Testing GPU functionality with tensor operations...\n")
    
    # Determine device (GPU if available, otherwise CPU)
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")
    
    if device.type == 'cuda':
        print(f"GPU Name: {torch.cuda.get_device_name(0)}\n")
    
    # Create random tensors
    print("Creating tensors...")
    matrix_size = 5000
    
    # Create two large matrices for multiplication
    tensor_a = torch.randn(matrix_size, matrix_size)
    tensor_b = torch.randn(matrix_size, matrix_size)
    
    print(f"Tensor A shape: {tensor_a.shape}")
    print(f"Tensor B shape: {tensor_b.shape}")
    
    # Move tensors to the device (GPU or CPU)
    print(f"\nMoving tensors to {device}...")
    tensor_a = tensor_a.to(device)
    tensor_b = tensor_b.to(device)
    
    # Perform matrix multiplication and measure time
    print(f"\nPerforming matrix multiplication on {device}...")
    start_time = time.time()
    
    result = torch.matmul(tensor_a, tensor_b)
    
    # Synchronize GPU operations (ensures timing is accurate)
    if device.type == 'cuda':
        torch.cuda.synchronize()
    
    end_time = time.time()
    elapsed_time = end_time - start_time
    
    print(f"\n✅ Matrix multiplication completed successfully!")
    print(f"Result shape: {result.shape}")
    print(f"⏱ Computation time: {elapsed_time:.4f} seconds")
    
    # Display a small sample of the result
    print(f"\nSample result (top-left 3x3 corner):")
    print(result[:3, :3].cpu().numpy())
    
    # Memory usage information (if on GPU)
    if device.type == 'cuda':
        allocated = torch.cuda.memory_allocated(0) / (1024**2)  # Convert to MB
        cached = torch.cuda.memory_reserved(0) / (1024**2)
        print(f"\n GPU Memory Usage:")
        print(f"  Allocated: {allocated:.2f} MB")
        print(f"  Cached: {cached:.2f} MB")
    
    print("\n GPU test completed successfully! Your environment is ready for deep learning.")
    
except Exception as e:
    print(f"❌ Error during GPU test: {e}")
    print("Please check your GPU configuration and try again.")
```
