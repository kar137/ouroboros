# Cross-Dataset Comparison

Compare CIFAR-10, Fashion-MNIST, and CIFAR-100 baselines.

```python
import sys
from pathlib import Path

candidate_roots = [
    Path('/content/drive/MyDrive/ouroboros/'),
]
project_root = next((p for p in candidate_roots if p.exists()), None)
if project_root is None:
    raise FileNotFoundError('Project root not found. Update candidate_roots.')
sys.path.insert(0, str(project_root))

import json
from src.metrics import plot_comparison_curves

metrics_paths = {
    'CIFAR-10': project_root / 'results/outputs' / 'cifar10_baseline_metrics.json',
    'Fashion-MNIST': project_root / 'results/outputs' / 'fashion_mnist_baseline_metrics.json',
    'CIFAR-100': project_root / 'results/outputs' / 'cifar100_baseline_metrics.json',
}

available = {name: path for name, path in metrics_paths.items() if path.exists()}
missing = [name for name, path in metrics_paths.items() if not path.exists()]
if missing:
    print('Missing metrics for:', missing)
    print('Run the baseline experiment notebooks first.')
else:
    comparison_plots = plot_comparison_curves(available, output_dir='plots', prefix='baseline_comparison')
    print('Saved comparison plots:', comparison_plots)
    summary = []
    for name, path in available.items():
        data = json.loads(path.read_text())
        last = data['epoch_metrics'][-1]
        summary.append({
            'dataset': name,
            'final_val_accuracy': last['validation']['accuracy'],
            'final_val_loss': last['validation']['loss'],
        })
    for row in summary:
        print(row)
```
